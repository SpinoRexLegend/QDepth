```python
raw_df = pd.read_csv('../Dataset/final_sdb_dataset_clean.csv')
print(f"Raw dataset shape: {raw_df.shape}")
raw_df.head()
```

```python
# Filter optical solar penetration zone (depth <= 30m)
sdb_df = raw_df[raw_df['depth'] <= 30.0].copy().reset_index(drop=True)
print(f"Filtered SDB dataset shape: {sdb_df.shape}")
sdb_df.describe()
```

```python
# Compute Stumpf log-ratio feature
sdb_df['stumpf_ratio'] = np.log(1000 * sdb_df['blue_band']) / np.log(1000 * sdb_df['green_band'])
sdb_df[['blue_band', 'green_band', 'stumpf_ratio', 'depth']].head()
```
